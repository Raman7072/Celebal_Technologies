```
from concurrent.futures import ThreadPoolExecutor

def run_notebook(notebook_name):
    dbutils.notebook.run(notebook_name, 0)

# List of notebooks to run
notebooks = ["Notebook-01", "Notebook-02", "Notebook-03", "Notebook-04"]

# Run notebooks in parallel
with ThreadPoolExecutor(max_workers=len(notebooks)) as executor:
    results = executor.map(run_notebook, notebooks)

# Check for results
for result in results:
    print(result)
```
