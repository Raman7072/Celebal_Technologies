Read from the source table

```
source_df = spark.read.format("delta").load("/FileStore/bronzedata")
source_df.display()
```

- Perform transformation or filtering if needed
- For example, let's filter for some condition

```
filtered_df = source_df.filter(source_df['tournament'] == 'Friendly')
filtered_df.display()
```

Insert/Update into Delta tables

```
filtered_df.write.format("delta").mode("append").saveAsTable("output_table_1")
```
