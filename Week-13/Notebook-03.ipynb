```
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```
source_df = spark.read.format("delta").load("/FileStore/bronzedata")
source_df.display()
```

```
source_df = source_df.withColumn("day", split(col("date"), "-").getItem(2))
source_df.display()
```

```
filtered_df = source_df.filter(source_df['day'] == '17')
filtered_df.display()
```

```
filtered_df.write.format("delta").mode("append").saveAsTable("output_table_3")
```
